```python
import pandas as pd
import subprocess
from pathlib import Path

# 1. Загружаем реестр
registry_file = "2026-05-07T11-14_export.csv"
df = pd.read_csv(registry_file)
df['Date'] = pd.to_datetime(df['Date'])
target_days = sorted(df['Date'].dt.date.unique())

# 2. Абсолютный путь для папки (Windows так надежнее)
output_dir = Path.cwd() / "data" / "eurnok_smart"
output_dir.mkdir(parents=True, exist_ok=True)

print(f"Запускаем жесткую загрузку ({len(target_days)} дней)...")

# 3. Цикл с subprocess
for i, date in enumerate(target_days, 1):
    start_date = date.strftime('%Y-%m-%d')
    end_date = (date + pd.Timedelta(days=1)).strftime('%Y-%m-%d')
    
    print(f"[{i}/{len(target_days)}] Загрузка {start_date}...", end=" ")
    
    # КРИТИЧЕСКОЕ ИСПРАВЛЕНИЕ: В Windows нужно использовать npx.cmd
    cmd = f"npx.cmd dukascopy-node -i eurnok -from {start_date} -to {end_date} -t tick -f csv -dir {output_dir}"
    
    try:
        # shell=True заставляет скрипт работать как реальный терминал
        result = subprocess.run(cmd, shell=True, check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
        print("Готово!")
    except subprocess.CalledProcessError as e:
        # Если будет ошибка, мы наконец-то увидим ее текст
        print("ОШИБКА!")
        print(e.stderr.decode('utf-8', errors='ignore'))
        break # Останавливаем цикл, чтобы не спамить ошибками

print("\nСкрипт завершен. Проверь папку data/eurnok_smart/")
```
